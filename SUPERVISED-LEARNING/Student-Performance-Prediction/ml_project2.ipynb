```
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt

# DATA LOAD KARYCHA
data = pd.read_csv("student_perf.csv")

# NaN values remove karycha
data = data.dropna()

# INPUT AND OUTPUT GHYAYCHA
x = data[['StudyHoursPerWeek']]
y = data['FinalGrade']

# MODEL TRAIN KARYCHA
model = LinearRegression()
model.fit(x, y)

predicted_score = model.predict(x)

# REGRESSION METRICS VALID AAHE KA BAGHYCHA
MAE = mean_absolute_error(y, predicted_score)
MSE = mean_squared_error(y, predicted_score)
RMSE = np.sqrt(MSE)
r2 = r2_score(y, predicted_score)

# RESULT DAKHVNAR
print("MAE:", round(MAE, 2))
print("MSE:", round(MSE, 2))
print("RMSE:", round(RMSE, 2))
print("R2:", round(r2, 4))  # closer to 1 = better

# HISTOGRAM
plt.figure(figsize=(10, 6))
plt.hist(data['FinalGrade'], bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram of Final Grade')
plt.xlabel('Final Grade')
plt.ylabel('NUmber of Students')
plt.show()

#regression line + scatterplot
plt.figure(figsize=(10, 6))
plt.scatter(x,y,color='blue',label='actual')
plt.plot(x,predicted_score,color='red',label='predicted(regression line)')
plt.title('model prediction vs actual score')
plt.xlabel('StudyHoursPerWeek')
plt.ylabel('final grade')
plt.legend()
plt.show()

new_hours = 9
predicted_new_score = model.predict([[new_hours]])
print(f"predicted final score for {new_hours} hours is{predicted_new_score} score")
```
