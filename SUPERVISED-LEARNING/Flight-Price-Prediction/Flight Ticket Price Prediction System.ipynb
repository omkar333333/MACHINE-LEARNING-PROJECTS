```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, recall_score, r2_score
from sklearn.preprocessing import OneHotEncoder

data=pd.read_csv("C:/Users/ROG STRIX/Desktop/project3/airlines_flights_data.csv")

print(pd.isnull(data).sum())

x=data[["airline",
        "source_city",
        "departure_time",
        "stops",
        "arrival_time",
        "destination_city",
        "class",
        "duration",
        "days_left"]]
y=data[["price"]]
```

## DATA PREPROCESSING
#### 1)METHOD_1

```python
x = pd.get_dummies(
    x,
    columns=[
        "airline",
        "source_city",
        "departure_time",
        "stops",
        "arrival_time",
        "destination_city",
        "class"
    ],
    dtype=int


)
print(x)
```

#### 2)METHOD_2

```python
x=data[["airline","source_city","departure_time","stops","arrival_time","destination_city","class","duration","days_left"]]

cat_col=["airline",
         "source_city",
         "departure_time",
         "stops",
         "arrival_time",
         "destination_city",
         "class"]

num_col=["duration",
         "days_left"]

ohe=OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore",
)

encoded=ohe.fit_transform(x[cat_col])

encoded_df=pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(cat_col)
)

print(encoded_df)

x_final=np.concatenate(
    [x[num_col],encoded_df],
    axis=1
)

print(x_final)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    x_final,y,train_size=0.8,test_size=0.2,random_state=42
)
```

```python
model=LinearRegression()
model.fit(X_train,Y_train)

y_pred=model.predict(X_test)
print(y_pred)

MSE=mean_squared_error(Y_test,y_pred)
print("MSE:",MSE)
MAE=mean_absolute_error(Y_test,y_pred)
print("MAE:",MAE)
RMSE=np.sqrt(mean_squared_error(Y_test,y_pred))
print("RMSE:",RMSE)
r2=r2_score(Y_test,y_pred)
print("r2:",r2)

```
